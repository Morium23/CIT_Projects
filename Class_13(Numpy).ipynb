# Lambda

```python
# Uses of lambda__1
def square(n):
    return n**2
print("Function: ",square(5))

square1 = lambda n:n**2
print("Lambda: ",square1(5))
```

```python
# Uses of lambda__2
def sum1(a,b):
    return a+b
print("Function: ",sum1(5,4))
sum2 = lambda a,b:a+b
print("Lambda: ",sum2(5,4))
```

# Array Comparison

```python
a = np.array([1,2,3])
b = np.array([5,2,6])
# check every element individually
print("Individual check: ",np.equal(a,b))
# check whole array
print("whole array check: ",np.array_equal(a,b))
```

# Aggregate Function

```python
a = [1,2,4]
print("Sum: ",np.sum(a))
print("Minimum: ",np.min(a))
print("Mean: ",np.mean(a))
print("Median: ",np.median(a))
print("Coorelaton Coefficient: ",np.corrcoef(a))
print("Standard Deviation: ",np.std(a))
```

# Array manipulation

```python
import numpy as np
e = np.array([1,2,3])
f = np.array([4,5,6])
print("Normal concatenate: ",np.concatenate((e,f)))
print("Horizontally: ",np.hstack((e,f)))       # --->
print("Vertically: ",np.vstack((e,f)))         # !!
print("Column wise: ",np.column_stack((e,f)))  # !!
```
