# Conditional Statement

```python
a,b = 4,10
if a<b:
    print('B is greater')
print('Always show')
```

```python
a,b = 4,10
if a<b:
    print('B is greater')
    print('Inside if')
```

```python
#if - checks all the conditions till the end
c,d = 10,11
if c<d:
    print("D is greatest")
if c>d:
    print('C is greatest')
if c == d:
    print('They are equal')
```

```python
#elif - (elseif)when got one condition true print it and terminate the code, not check the rest conditions
c,d = 10,11
if c<d:
    print("D is greatest")
elif c>d:
    print('C is greatest')
elif c >= d:
    print('They are equal')
```
