```python
#50  Write a Python program to sort a list of nested dictionaries.
my_list = [{'key':{'subkey':1}},{'key':{'subkey':10}},{'key':{'subkey':5}}]
ext = []
for i in my_list:
    ext.append(i['key']['subkey'])
ext = sorted(ext,reverse = True)
print(ext)
final = []
for i in ext:
    for j in my_list:
        if j['key']['subkey'] == i:
            final.append(j)
final
```
