# Visualization library
# lineplot, scatterplot, barplot, pieplot

```python
import matplotlib.pyplot as plt
```

# Line plot

```python
X = list(range(1,100,10))
Y = list(range(1,50,5))
plt.plot(X,Y)
plt.xlabel('Range: 1-100')
plt.ylabel('Range 1-50')
plt.title("Get line plot")
plt.show()
```

```python
X = list(range(0,101,10))
Y = list(range(0,51,5))
plt.plot(X,Y, linewidth = 2.0, linestyle = ":",
         color = 'red', alpha = 0.9, marker = 'o')
# alpha - brigthness, marker-(4,x,0) - points
plt.xlabel('Range: 1-100')
plt.ylabel('Range 1-50')
plt.title("Get line plot")
plt.show()
```

```python
X1 = [2,4,6,8]
Y1 = [6,8,4,6]
plt.plot(X1,Y1)
plt.show()
```

```python
X1 = [2,4,6,8]
Y1 = [6,8,4,6]
Z1 = [6,9,3,10]
plt.plot(X1,Y1, color='red')
plt.plot(Y1,Z1, color='green')
plt.legend(['Graph-1',"Graph-2"])
plt.show()
```

# Scatter plot

```python
X2 = [2,4,6,8]
Y2 = [6,8,4,6]
plt.scatter(X2,Y2, color='red')
plt.show()
```

```python
# Extra feature
X2 = [2,4,6,8]
Y2 = [6,8,4,6]
plt.scatter(X2,Y2, color='red', s=200, edgecolor='y',
            marker='4', alpha=1)
plt.show()
```

```python
X1 = [2,4,6,8]
Y1 = [6,8,4,6]
Z1 = [6,9,3,10]
plt.scatter(X1,Y1, color='red')
plt.scatter(Y1,Z1, color='green')
plt.legend(['Graph-1',"Graph-2"])
plt.grid()
plt.show()
```

# Bar plot

```python
X2 = ['mango','orange','banana']
Y2 = [5,12,15]
plt.xlabel("Fruits name:")
plt.ylabel("Numbers:")
plt.title("Distribution of fruits:")
plt.bar(X2,Y2, color='red')
plt.show()
```

```python
# horizontal representation
X2 = ['mango','orange','banana']
Y2 = [5,12,15]
plt.xlabel("Fruits name:")
plt.ylabel("Numbers:")
plt.title("Distribution of fruits:")
plt.barh(X2,Y2, color='red')
plt.show()
```

# Subplot

```python
X1 = [2,4,6,8]
Y1 = [6,8,4,6]
Z1 = [6,9,3,10]

plt.subplot(2,1,1)
plt.plot(X1,Y1, color='red')
plt.legend(['Graph-1'])
plt.grid()

plt.subplot(2,1,2)
plt.plot(Y1,Z1, color='green')
plt.legend(["Graph-2"])

plt.grid()
plt.show()
```

# Piechart

```python
import matplotlib.pyplot as plt
%matplotlib inline

lab = ['Dog','Cat','Wolf','Lion']
sizes = [20,30,30,80]

#creating pie chart
plt.pie(sizes,labels = lab,autopct='%1.2f%%',
       shadow=True, startangle=90,
        explode=(0,0,0,0))
#labels - words that tag outside the circle
# autopct - the numbers written inside the circle
            #.2 means 2 numbers after . like 2.45
```
