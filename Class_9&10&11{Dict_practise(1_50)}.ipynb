```python
#19 Write a Python program to combine two dictionary adding values for common
# keys.
d1 = {'a': 100, 'b': 200, 'c':300}
d2 = {'a': 300, 'b': 200, 'd':400}
dict19 = {}
for i in d1:
    for j in d2:
        if i == j:
            dict19[i] = d1.get(i) + d2.get(j)
        else:
            dict19[i] = d1.get(i)
            dict19[j] = d2.get(j)
dict19
```
