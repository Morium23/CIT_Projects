# Merge

```python
import pandas as pd
df1=pd.DataFrame({'ID':[1,2,3,4],
                'Name':['Alice','Bob','Charlie','David']})
df2=pd.DataFrame({'ID':[3,4,5,6],
                'Age':[25,30,35,40]})
df1
```

```python
df2
```

```python
pd.merge(df1,df2,on = 'ID')
```

```python
pd.merge(df1,df2,on = 'ID', how='inner')   #by default
```

```python
pd.merge(df1,df2,on = 'ID', how='outer') # combine
```

```python
pd.merge(df1,df2,on = 'ID', how='left') #prioritized left
```

```python
pd.merge(df1,df2,on = 'ID', how='right')
```

# join

```python
import pandas as pd
df3=pd.DataFrame({'ID':[1,2,3],
                'Name':['A','B','C']},index=['a','b','c'])
df4=pd.DataFrame({'ID':[4,5,6],
                'Age':[30,35,40]},index=['b','c','d'])
df3
```

```python
df4
```

```python
joined_df = df3.join(df4, how='outer')
```
