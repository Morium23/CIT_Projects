```python
# with parameter
def arithmetic_operation(a,b,operator):
    if operator=='+':
        c = a+b
    elif operator == '-':
        c = a-b
    elif operator == '*':
        c = a*b
    elif operator == '/':
        c = a/b
    print("The result is ",c)
    return c
result = arithmetic_operation(8,4,"*")
result
```

```python
def check_prime(n):
    initial = "prime"
    for element in range(2,n):
        if n%element == 0:
            initial = "non-prime"
            break
    #print("The value of ",n,"is ",initial)
    return initial
```

```python
result = check_prime(5)
result
```

# While loop

```python
#print all the prime numbers from the range 1,100
count = 1
i = 2
while(count<=100):
    result = check_prime(i)
    if result=='prime':
        print(i)
    count +=1
    i+=1
```

```python
# define a class using object(without parameter)
class only_subtract:
    def subtract(self):
        a = 10
        b = 4
        return a-b
sub_obj = only_subtract()
sub_obj.subtract()
```

```python
# define a class using object(with parameter)
class only_sum:
    def summ(self,a,b):
        return a+b
only_obj = only_sum()
only_obj.summ(5,6)
```

```python
# uses of self in variable
class car_feature:
    def features(self,number_plate):
        car_name = 'Toyota'
        car_color = 'ash'
        self.num_plate = number_plate
        print(car_name," ",car_color," ",self.num_plate)
    def display(self):
        print(self.num_plate)
car_obj = car_feature()
car_obj.features('AB2014')
car_obj.display()
```

```python
# Constructor
class rickshaw:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def display(self):
        print("Name: ",self.name,"& Age: ",self.age)
rick_obj = rickshaw("Mizan",26)
rick_obj.display()
```

```python
# Calling a function under another one
class Calculator:
    def even_odd(self,a):
        if a%2 == 0:
            result = self.square(a)
            print("The number",a," is even & result is ",result)
        else:
            result = self.double(a)
            print("The number",a," is odd & result is ",result)
    def square(self,b):
        return b**2
    def double(self,c):
        return c*2
cal_obj = Calculator()
cal_obj.even_odd(6)
cal_obj.even_odd(7)
```
