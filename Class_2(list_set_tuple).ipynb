```python
set_1={'i','n','t','e','l'}
set_2={'p','y','t','h','o'}
set_1.union(set_2)
```

```python
set_1.intersection(set_2)
```

```python
set_1.difference(set_2)    #set_1 - set_2
```

```python
set_2.difference(set_1)    #set_2 - set_1
```
